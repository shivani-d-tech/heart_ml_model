# 2. Data Cleaning
I decided to carry out data cleaning before EDA for this particular dataset as most columns were categorical data but represented as a float. Therefore, I converted these columns in order to aid in smoother analysis during EDA.

### Removal of rows with missing values

```python
#As we know '?' is only in 'thal' and 'ca' columns can subset the dataframe to exclude these
df = df[(df['ca'] != '?') & (df['thal'] != '?')]

#Check this worked
print((df == '?').sum())
```

### Change Encoded Columns to Categorical Data Types

```python
#Change 'thal' and 'ca' to numeric columns as they were already strings
#Do this so they are able to be converted into integers (easier for preprocessing step)
df[['thal','ca']] = df[['thal','ca']].apply(pd.to_numeric)

#Turn encoded columns into 'object' data type
update_cols = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'thal', 'ca', 'target']
df[update_cols] = df[update_cols].astype('Int64')
df[update_cols] =  df[update_cols].astype('object')

#Check that this has worked
df.dtypes
```

```python
#Preview data
df.head()
```

## Summary: Data Cleaning

* First, we removed all rows containing '?' in `thal` and `ca`. The presence of '?' was blocking these columns from conversion into integer format to remove the '.0'.
* we have converted all encoded columns into integers (to get rid of '.0') and then turned them into categorical variables ready for EDA and preprocessing.
* The data is now ready for EDA and preprocessing
